# Test of hypothesis

Question:
Is the average footlenght of male significantly different from female?

Ho: Mean footleght of a male possum = mean footleght of a female possum
H1: Mean footleght of a male possum != mean footleght of a female possum

```python
male_possum = df["sex"] == "m"
female_possum = df["sex"] == "f"
```

```python
df.head()
```

```python
df['sex'].head()
```

```python
male_df = df[male_possum]
female_df = df[female_possum]
```

```python
male_mean = male_df['footlgth'].mean()
```

```python
female_mean = female_df['footlgth'].mean()
```

```python
from scipy import stats
```

```python
np.random.seed(42)
```

```python
female_sample = np.random.normal(loc=female_mean, scale=5, size=30)
male_sample = np.random.normal(loc=male_mean, scale=5, size=30)
```

```python
female_sample
```

```python
male_sample
```

```python
# T-Test
t_stat, p_val = stats.ttest_ind(female_sample, male_sample)
```

```python
print(f"T-statistics: {t_stat}")
```

```python
print(f"P-value: {p_val}")
```
