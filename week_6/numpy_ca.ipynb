```python
# Q1 create a NumPy array containing integers from 1 to 20. Display the array and its shape, size, and ndim.
array = np.arange(1,21)
print(array.shape)
print(array.ndim)
print(array.size)
```

```python
# Q2. Create a 2D array (3×4) of random integers between 10 and 99. Display the array and print its dtype, itemsize, and nbytes.
array = np.random.randint(10,100,(3,4))
print(array)
print("type:", array.dtype)
print("size:", array.size)
print("nbytes:", array.nbytes)
```

```python
# Q3. Generate the following special arrays: A 5×5 array of zeros, A 4×4 array of ones, A 6×6 identity matrix.
array = np.zeros((5,5))
array2 = np.ones((4,4))
array3 = np.identity(6)
print(array)
print('\n')
print(array2)
print('\n')
print(array3)
```

```python
# Q4. Use np.arange to create an array of integers from 0 to 50 with a step of 5. Then use np.linspace to generate 10 evenly spaced values between 0 and 1.
np.arange(0,51,5)
np.linspace(0,1,10)
```
