### Task 6

```python
def square_of_numbers(arr):
    square = arr.copy()
    square_no = np.where(square % 2 == 0, square ** 2, -1)
    return square_no
```

```python

rows = int(input("Enter the number of rows: "))
columns = int(input("Enter the number of columns: "))

matrix = []
for i in range(rows):
    row = []
    for j in range(columns):
        element = int(input(f"Enter element [{i}][{j}]"))
        row.append(element)
    matrix.append(row)
```

```python
arr = np.array(matrix)
new_arr = square_of_numbers(arr)

print("User Matrx\n")
print(arr)
print("\nSquare of even elemet in user Matrix with odd numbers replaced")
print(new_arr)
```
