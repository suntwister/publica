```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import duckdb
```

```python
file_path = "../../data/Crimes.csv"
```

```python
df = duckdb.query(f"""
    SELECT * FROM read_csv_auto('{file_path}')                  
""").to_df()
```

```python
df.info()
```

```python
print("Rows, Columns:\n", df.shape)
```

```python
df.isna().sum()
```

```python
df.columns
```

```python
if "Latitude" in df.columns and "Longitude" in df.columns:
    df = df.dropna(subset=["Latitude", "Longitude"])
```

```python
df.isna().sum()
```

```python
df = df.dropna()
```

```python
df = df.drop(columns=["ID"])
```

```python
# Convert 'Date' column to datetime
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Convert Arrest/Domestic columns to boolean if needed
if "Arrest" in df.columns:
    df["Arrest"] = df["Arrest"].astype(bool)
if "Domestic" in df.columns:
    df["Domestic"] = df["Domestic"].astype(bool)

# Standardize crime type
if "Primary_Type" in df.columns:
    df["Primary_Type"] = df["Primary_Type"].str.title().str.strip()

```

```python
# Drop duplicates
df = df.drop_duplicates()
```

```python
df.to_parquet("../../data/Crimes_clean2.parquet", index=False, engine="fastparquet", compression="snappy")
```
