```python
import numpy as np
```

```python
A = np.array(
    [
        [2,-1,3],
        [0,4,1]
    ])

B = np.array(
    [
        [1,2],
        [-2,0],
        [5,-3]
    ]
)
```

```python
print(f"The dimension of Matrix A is {A.shape}")
print(f"The dimension of Matrix B is {B.shape}")
```

```python
A_row, A_col = A.shape
B_row, B_col = B.shape
```

```python
print(f"The number of rows in Matrix A is {A_row}")
print(f"The number of columns in Matrix A is {A_col}")
```

```python
print(f"The number of rows in Matrix B is {B_row}")
print(f"The number of columns in Matrix B is {B_col}")
```

** Matrix A can multiply Matrix B since the column of A is equal row of B, resulting to a (2,2) Matrix

```python
A_col == B_row
```

Multiplication of Matrix A and B

```python
AB = np.dot(A,B)
print("Multiplication of Matrix A and B")
print(AB)
```

** Matrix B can also multiply Matrix A since the column of B is equal row of A, resulting to a (3,3) Matrix

```python
B_col == A_row
```

```python
BA = np.dot(B,A)
print("Multiplication of Matrix B and A")
print(BA)
```
